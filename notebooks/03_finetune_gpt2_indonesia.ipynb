# Text Generation Bahasa Indonesia: Eksplorasi & Fine-tuning

Berbeda dari notebook Code Generation (train GPT-2 dari nol), di sini kita:

```python
import torch
from transformers import GPT2Tokenizer, GPT2LMHeadModel, pipeline
```

## Bagian A — Eksplorasi Model Pretrained Bahasa Indonesia

### A.1 Fill-Mask (Masked Language Model)

BERT dan RoBERTa bukan model generatif murni — tugasnya menebak kata yang ditutup (`[MASK]`/`<mask>`), bukan melanjutkan teks.

```python
unmasker = pipeline('fill-mask', model='cahya/bert-base-indonesian-522M')
unmasker("Ibu ku sedang bekerja [MASK] supermarket")
```

```python
unmasker = pipeline('fill-mask', model='cahya/roberta-base-indonesian-522M')
unmasker("Ibu ku sedang bekerja <mask> supermarket")
```

### A.2 Text Generation dengan GPT-2 Indonesia

Dokumentasi: https://github.com/cahya-wirawan/indonesian-language-models/tree/master/Transformers

#### GPT-2 Small Indonesian

```python
tokenizer = GPT2Tokenizer.from_pretrained("cahya/gpt2-small-indonesian-522M")

# add the EOS token as PAD token to avoid warnings
model = GPT2LMHeadModel.from_pretrained("cahya/gpt2-small-indonesian-522M", pad_token_id=tokenizer.eos_token_id)
```

```python
input_sentences = [
    'Alkisah pada jaman dahulu kala seekor sapi tengah melintas di sebuah hutan belantara. Sapi hutan itu sedang merasa kehausan di tengah panasnya terik matahari',
    'Cirebon adalah sebuah kota kecil di Jawa Barat dengan keindahannya yang melebihi kota Bandung',
    'Sriwijaya adalah salah satu kemaharajaan bahari yang pernah berdiri di pulau Sumatra dan banyak memberi pengaruh di Nusantara dengan daerah kekuasaan yang luas',
    'Pantai berpasir putih ini cukup populer akhir-akhir ini karena menawarkan pemandangan yang begitu eksotis, indah dan mempesona',
]
```

```python
# We set the seed manualy for reproducible result
for i, sentence in enumerate(input_sentences):
    torch.manual_seed(1)
    input_ids = tokenizer.encode(sentence, return_tensors='pt')
    sample_output = model.generate(
        input_ids,
        do_sample=True,
        max_length=150,
        top_k=50,
        top_p=0.95,
    )
    print("Output:\n" + 100 * '-')
    print("{}: {}".format(i, tokenizer.decode(sample_output[0], skip_special_tokens=True)))
```

#### GPT-2 Large Indonesian

```python
tokenizer = GPT2Tokenizer.from_pretrained("cahya/gpt2-large-indonesian-522M")

# add the EOS token as PAD token to avoid warnings
model = GPT2LMHeadModel.from_pretrained("cahya/gpt2-large-indonesian-522M", pad_token_id=tokenizer.eos_token_id)
```

## Bagian B — Fine-tuning GPT-2 Indonesia pada Dataset Sendiri

Dokumentasi model dasar: https://huggingface.co/cahya/gpt2-small-indonesian-522M

Alur: load model pretrained → siapkan dataset kecil (dummy) → tokenisasi → training dengan `Trainer` → simpan model → uji generate.

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel

tokenizer = GPT2Tokenizer.from_pretrained("cahya/gpt2-small-indonesian-522M")

# Set the pad token for the tokenizer, as it's required for batching
tokenizer.pad_token = tokenizer.eos_token

# add the EOS token as PAD token to avoid warnings
model = GPT2LMHeadModel.from_pretrained("cahya/gpt2-small-indonesian-522M", pad_token_id=tokenizer.eos_token_id)
```

### B.2 Training Arguments

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=10,          # dataset dummy kecil, epoch bisa dinaikkan; untuk dataset asli sesuaikan
    per_device_train_batch_size=2,  # sesuaikan dengan memori GPU
    save_steps=10_000,
    save_total_limit=2,
    logging_steps=500,
    learning_rate=2e-5,
    weight_decay=0.01,
)
```

### B.3 Training & Simpan Model

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    eval_dataset=tokenized_dataset,  # dataset yang sama dipakai untuk eval, hanya untuk demo
    data_collator=data_collator,
)

trainer.train()
```

```python
# Save the fine-tuned model
model_output_path = "./fine_tuned_gpt2"
trainer.save_model(model_output_path)
tokenizer.save_pretrained(model_output_path)

print(f"Model and tokenizer saved to {model_output_path}")
```

### B.4 Uji Coba Generate

```python
from transformers import pipeline

# Load the fine-tuned model and tokenizer
generator = pipeline(
    'text-generation',
    model=model_output_path,
    tokenizer=model_output_path,
)

prompt = "Trex merupakan"
generated_text = generator(prompt, max_length=50, num_return_sequences=1)

print(f"Prompt: {prompt}")
print(f"Generated text: {generated_text[0]['generated_text']}")
```
